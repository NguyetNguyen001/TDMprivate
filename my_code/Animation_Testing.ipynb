```python
import CalculatedFieldSubroutines as cfs

#

import numpy as np

import pandas as pd

#

import matplotlib.pyplot as plt

from matplotlib.animation import FuncAnimation

from matplotlib.animation import FFMpegWriter

#

from pandasgui import show

#

import warnings
```

```python
#



fig, ( ax1, ax2 ) = plt.subplots( 1, 2 )

#

t_points = np.linspace( 0, 2 * np.pi, 1000 )

y_points = np.sin( t_points )

z_points = np.cos( t_points )

#

ax1.set_xlim( min( t_points ), max( t_points ) )

ax1.set_ylim( min( y_points ), max( y_points ) )

ax2.set_xlim( min( t_points ), max( t_points ) )

ax2.set_ylim( min( z_points ), max( z_points ) )

#

animated_plot1, = ax1.plot( [], [], marker = 'o', ls = '' )

animated_plot2, = ax2.plot( [], [], marker = 'o', ls = '' )

def update( frame_num ):
    
    animated_plot1.set_data( [ t_points[ frame_num ] ], [ y_points[ frame_num ] ] )

    animated_plot2.set_data( [ t_points[ frame_num ] ], [ z_points[ frame_num ] ] )

#

animation = FuncAnimation( fig = fig, func = update, frames = len( t_points ) )

#

animation.save( 'test.mp4', writer = FFMpegWriter( fps = len( t_points ) / 10 ) )

#

plt.close()
```
