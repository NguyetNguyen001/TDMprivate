```python
import CalculatedFieldSubroutines as cfs

#

import numpy as np

import pandas as pd

#

import os
```

```python
gmIDs = cfs.list_whitelisted_gmIDs()

topics = cfs.list_topics()

print( topics )
```

```python
def CreatePreprocessedMovingDataFolder( moving_window ): # sec
    
    for gmID in gmIDs:

        df = cfs.retrieve_gmID_preprocessed_moving_data_v3( gmID = gmID, moving_window = moving_window )

        #

        df = cfs.df_undersampler( df = df, 
                                  colname_to_undersample_by = 'time', 
                                  target_diff = 0.25 * 1e9, #ns 
                                  colname_vals_to_except = { 'BinaryDisengagement' : 1 } )

        #

        filepath = f'{ cfs.origin_dir() }/Preprocessed_Moving_Data_v3_reduced/{ moving_window }sec_moving_window/{ gmID }'

        os.makedirs( filepath, exist_ok = True )

        #

        df.to_csv( f'{ filepath }/{ gmID }.csv', index = False )
```

```python
for sec in [ 2, 4, 0 ]:

    CreatePreprocessedMovingDataFolder( moving_window = sec )

    print( f'{ sec } second moving window data done' )
```
